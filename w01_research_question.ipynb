## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*
This lane is clearly worth pursuing...
54.2% of pages in the starter dataset are currently declining, a substantial share of the inventory. However, my specific "stale + visible" rule from Week 1 (180+ days since update, 500+ impressions) only catches 17 pages (0.1%) far too narrow to represent the real opportunity. This tells me the threshold itself needs work over the coming weeks: the problem is much bigger than this one rule currently detects, and refining what counts as "worth reviewing" will be a core part of the capstone.

```python
import os
print("Current folder:", os.getcwd())
print("Contents:", os.listdir("."))
# Load the starter dataset — same one used in Notebooks 01 and 02
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: how many pages are stale AND still getting real traffic?
stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
print(f"Stale + visible pages: {len(stale_visible)} out of {len(df)} total ({len(stale_visible)/len(df)*100:.1f}%)")

# Number 2: what share of all pages are currently declining?
declining_pct = (df["trend_direction"] == "down").mean() * 100
print(f"Pages currently declining: {declining_pct:.1f}%")

# Number 3: how much impression volume is sitting in stale+visible pages?
total_impressions = df["impressions_90d"].sum()
stale_visible_impressions = stale_visible["impressions_90d"].sum()
print(f"Impressions tied up in stale+visible pages: {stale_visible_impressions:,.0f} "
      f"({stale_visible_impressions/total_impressions*100:.1f}% of all impressions)")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) and what it never will (causal proof, 'predicting Google').*
What I can claim: My work will be observational and decision-support only. I can say things like "this page shows signs consistent with decline" or "this pattern is associated with lower engagement" — based on what the data shows, not proof of cause. My ranked list is meant to help a human reviewer prioritize their limited time, not replace their judgment.

What I can't claim: I cannot prove that refreshing a page causes it to recover that would require an actual experiment (like A/B testing before/after a refresh), which this data doesn't provide. I also can't claim to have discovered anything about how Google's search algorithm actually works, I'm only observing patterns in FlyRank's own tracked data. Any drop in a page's performance could also be explained by seasonality, a related page absorbing its traffic, or plain noise not necessarily "decline" so I'll rule those out before calling something a real decline.
